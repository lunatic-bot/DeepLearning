### ReLU (Rectified Linear Unit)

ReLU is one of the most widely used activation functions in deep learning.

It helps neural networks learn complex patterns efficiently and solves many problems caused by sigmoid/tanh activations.

Definition

ReLU function:

f(x)=max(0,x)

Meaning:

if x>0 → output = x
if x≤0 → output = 0

### Key advantages of ReLU
1. Solves Vanishing Gradient Problem

For positive values:

Derivative = 1

So gradients remain strong during backpropagation.

Unlike sigmoid whose derivatives become tiny.

2. Faster Training

ReLU is computationally simple:

max(0, x)

No expensive exponential calculations like sigmoid.

So training becomes much faster.


3. Sparse Activation

Negative neurons output zero.

This creates sparse networks which are:

efficient
memory friendly
less prone to overfitting

### Disadvantage: Dying ReLU Problem

If neurons always receive negative inputs:

Output = 0 forever

Gradient also becomes zero.

Neuron stops learning.

This is called:

Dying ReLU
