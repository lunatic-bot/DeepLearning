### Regularization in Deep Learning

Regularization techniques are used to reduce overfitting in machine learning and deep learning models.

Regularization is a technique used to reduce overfitting by adding penalties or constraints to model complexity, improving generalization on unseen data.

Overfitting means:

model performs very well on training data
but poorly on unseen/test data

Regularization helps the model generalize better.

What is Regularization?

Regularization adds constraints/penalties so the model avoids becoming too complex.

## Dropout

Dropout is a special regularization technique used in neural networks.

During training, dropout randomly disables some neurons.

Dropout is a neural network regularization technique where random neurons are temporarily deactivated during training to prevent over-dependence among neurons and reduce overfitting.

Dropout = 0.5 means 50% neurons randomly turned off during each training step

Why does dropout help?

Without dropout:

neurons become overly dependent on each other
network memorizes training data

With dropout:

network learns robust independent features
reduces co-adaptation
improves generalization
