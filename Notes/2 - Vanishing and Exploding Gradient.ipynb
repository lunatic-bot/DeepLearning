#### Vanishing gradient problem
The Vanishing Gradient Problem is a common issue in deep neural networks where gradients become extremely small during backpropagation, causing earlier layers to learn very slowly or stop learning entirely.


Vanishing gradient is a problem in deep neural networks where gradients become extremely small during backpropagation, causing earlier layers to learn very slowly or not at all. It commonly occurs with sigmoid/tanh activations and is mitigated using ReLU, batch normalization, residual connections, and better initialization techniques.

Where it mostly occurs

Especially common in:

- Deep Neural Networks (DNNs)
- Recurrent Neural Networks (RNNs)

with activation functions like: Sigmoid  Tanh


#### Why sigmoid causes vanishing gradients

Sigmoid output range:

σ(x)=
1/1+e
−x

Its derivative:

σ
′
(x)=σ(x)(1−σ(x))

Maximum derivative is only:

0.25

Multiplying many such small numbers → gradient vanishes.

#### Solutions
1. ReLU activation (most important)

Instead of sigmoid/tanh use:

f(x)=max(0,x)

Why better?

Gradient is 1 for positive values
Prevents shrinking gradients

Variants:

Leaky ReLU
ELU
GELU

2. Proper weight initialization

3. Batch Normalization

4. Residual Connections (ResNet)

Skip connections allow gradients to flow directly.

5. LSTM / GRU (for RNNs)

Traditional RNNs heavily suffer from vanishing gradients.

LSTM and GRU use gating mechanisms to preserve long-term dependencies.

### Exploding gradient problem 

The Exploding Gradient Problem occurs when gradients become extremely large during backpropagation, causing unstable training and huge weight updates in deep neural networks.

It is essentially the opposite of the vanishing gradient problem.


Exploding gradient is a problem in deep neural networks where gradients grow exponentially during backpropagation, causing extremely large weight updates, unstable training, and failure to converge. It is commonly addressed using gradient clipping, proper initialization, batch normalization, and improved architectures like LSTM and ResNet.

Intuition

During backpropagation:

∂W/∂L=g1 × g2 × g3× ⋯

If gradients are repeatedly multiplied by values greater than 1, they grow exponentially.

Example:
5 × 5 × 5 × 5 = 625
Eventually gradients become enormous.

Where it commonly occurs

Especially in:

Deep Neural Networks
Recurrent Neural Networks (RNNs)
Very deep architectures

and when using:

large learning rates
poor initialization

Solutions
1. Gradient Clipping (most common)

Cap gradient magnitude to a threshold.
torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

This prevents gradients from becoming too large.


2. Proper Weight Initialization

Use:

Xavier Initialization
He Initialization

to maintain stable gradient flow.

3. Lower Learning Rate

Too high learning rate amplifies exploding gradients.

4. Batch Normalization

Helps stabilize activations and gradients.

5. Use LSTM / GRU instead of vanilla RNN

Traditional RNNs are very vulnerable to exploding gradients.
