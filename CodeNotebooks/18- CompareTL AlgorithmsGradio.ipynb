```python
import tensorflow as tf
import numpy as np
import requests
import gradio as gr
from PIL import Image


inception_net = tf.keras.applications.InceptionV3()
mobile_net = tf.keras.applications.MobileNetV2()

## download human redable labels
response = requests.get("https://git.io/JJkYN")
labels = response.text.split("\n")
print(labels)
```

```python
def classify_image_with_mobile_net(img):
    img = Image.fromarray(img.astype(np.uint8), 'RGB')
    img = img.resize((299, 299))
    img = np.array(img)
    img = img.reshape((-1, 299, 299, 3))
    img = tf.keras.applications.mobilenet.preprocess_input(img)  # handles scaling internally
    prediction = inception_net.predict(img).flatten()
    return {labels[i]: float(prediction[i]) for i in range(1000)}
```

```python
def classify_image_with_inception_net(img):
    img = Image.fromarray(img.astype(np.uint8), 'RGB')
    img = img.resize((299, 299))
    img = np.array(img)
    img = img.reshape((-1, 299, 299, 3))
    img = tf.keras.applications.inception_v3.preprocess_input(img)  # handles scaling internally
    prediction = inception_net.predict(img).flatten()
    return {labels[i]: float(prediction[i]) for i in range(1000)}
```

```python
imagein = gr.Image()
# label = gr.Label(num_top_classes=3)
sample_images=[
    ["images\monkey.jpg"],
    ["images\sailboat.jpg"],
    ["images\bicycle.jpg.jpg"],
    ["images\dog.jpg"],
    ["images\fox.jpg"],
]
```

```python
# gr.Interface([classify_image_with_mobile_net, classify_image_with_inception_net], imagein, label,
#              title="MobileNet and InceptionV3",
#              description="Compare the predictions of MobileNet and InceptionV3").launch()


def classify_both(img):
    mobile_result = classify_image_with_mobile_net(img)
    inception_result = classify_image_with_inception_net(img)
    return mobile_result, inception_result

label1 = gr.Label(num_top_classes=3, label="MobileNet")
label2 = gr.Label(num_top_classes=3, label="InceptionV3")

gr.Interface(
    fn=classify_both,
    inputs=imagein,
    outputs=[label1, label2],
    title="MobileNet and InceptionV3",
    description="Compare the predictions of MobileNet and InceptionV3"
).launch()
```
