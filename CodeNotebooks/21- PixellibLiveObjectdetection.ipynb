```python
import pixellib
from pixellib.instance import instance_segmentation
import cv2
import matplotlib.pyplot as plt
```

```python
segment_image = instance_segmentation()
segment_image.load_model(r"D:\Coding\DeepLearning\Models\mask_rcnn_coco.h5") #r"D:\Coding\DeepLearning\Models/mask_rcnn_coco.h5")

camera = cv2.VideoCapture(0)

while camera.isOpened():
    _, frame = camera.read()
    ## Apply the segmentation
    frame = segment_image.segmentFrame(frame, show_bboxes=True)
    image = frame[1]
    cv2.imshow("Image Segmenatation", image)

    if cv2.waitKey(10) & 0xFF == ord('q'):
        break

camera.release()
cv2.destroyAllWindows()
```
